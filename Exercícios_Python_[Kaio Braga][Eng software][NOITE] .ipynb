### 21. Busca Binária

Enunciado: Implemente busca binária para encontrar um elemento em uma lista ordenada.

Explicação: Divide a área de busca pela metade até encontrar o valor.

Código:


```python
def busca_binaria(lista, alvo):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2

        if lista[meio] == alvo:
            return meio
        elif lista[meio] < alvo:
            inicio = meio + 1
        else:
            fim = meio - 1

    return -1

numeros = [2, 5, 8, 12, 16, 23, 38, 56]
alvo = int(input("Digite o número que deseja buscar: "))

posicao = busca_binaria(numeros, alvo)

if posicao != -1:
    print("Número encontrado no índice", posicao)
else:
    print("Número não encontrado.")
```

### 22. Criptografia de Cifra de César

Enunciado: Crie uma função para criptografar uma mensagem usando a cifra de César.

Explicação: Desloca cada letra algumas posições no alfabeto.

Código:


```python
def criptografar_cesar(texto, deslocamento):
    resultado = ""

    for caractere in texto:
        if caractere.isalpha():
            base = ord("A") if caractere.isupper() else ord("a")
            novo = (ord(caractere) - base + deslocamento) % 26 + base
            resultado += chr(novo)
        else:
            resultado += caractere

    return resultado

mensagem = input("Digite a mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

print("Mensagem criptografada:", criptografar_cesar(mensagem, deslocamento))
```

Enunciado: Crie uma função para descriptografar uma mensagem cifrada com César.

Explicação: Faz o deslocamento contrário para recuperar a mensagem.

Código:


```python
def descriptografar_cesar(texto, deslocamento):
    resultado = ""

    for caractere in texto:
        if caractere.isalpha():
            base = ord("A") if caractere.isupper() else ord("a")
            novo = (ord(caractere) - base - deslocamento) % 26 + base
            resultado += chr(novo)
        else:
            resultado += caractere

    return resultado

mensagem = input("Digite a mensagem criptografada: ")
deslocamento = int(input("Digite o deslocamento utilizado: "))

print("Mensagem original:", descriptografar_cesar(mensagem, deslocamento))
```

### 24. Validação de CPF

Enunciado: Implemente uma função para validar um CPF brasileiro.

Explicação: Confere o formato e calcula os dois dígitos verificadores do CPF.

Código:


```python
def validar_cpf(cpf):
    cpf = "".join(caractere for caractere in cpf if caractere.isdigit())

    if len(cpf) != 11:
        return False

    if cpf == cpf[0] * 11:
        return False

    soma1 = sum(int(cpf[i]) * (10 - i) for i in range(9))
    resto1 = soma1 % 11
    digito1 = 0 if resto1 < 2 else 11 - resto1

    soma2 = sum(int(cpf[i]) * (11 - i) for i in range(10))
    resto2 = soma2 % 11
    digito2 = 0 if resto2 < 2 else 11 - resto2

    return cpf[-2:] == f"{digito1}{digito2}"

cpf = input("Digite o CPF: ")

if validar_cpf(cpf):
    print("CPF válido.")
else:
    print("CPF inválido.")
```

```python
palavra = "python"
letras_descobertas = set()
tentativas_erradas = 0
limite_erros = 6

while tentativas_erradas < limite_erros:
    exibicao = ""

    for letra in palavra:
        if letra in letras_descobertas:
            exibicao += letra + " "
        else:
            exibicao += "_ "

    print("\nPalavra:", exibicao)

    if all(letra in letras_descobertas for letra in palavra):
        print("Parabéns! Você venceu.")
        break

    tentativa = input("Digite uma letra: ").lower()

    if len(tentativa) != 1 or not tentativa.isalpha():
        print("Digite apenas uma letra.")
        continue

    if tentativa in palavra:
        letras_descobertas.add(tentativa)
    else:
        tentativas_erradas += 1
        print(f"Erro! Restam {limite_erros - tentativas_erradas} tentativa(s).")
else:
    print("Você perdeu. A palavra era:", palavra)
```

### 31. Ordenação por Bolha

Enunciado: Implemente Bubble Sort.

Explicação: Compara valores vizinhos e troca os que estão fora de ordem.

Código:


```python
def bubble_sort(lista):
    resultado = lista.copy()

    for fim in range(len(resultado) - 1, 0, -1):
        houve_troca = False

        for i in range(fim):
            if resultado[i] > resultado[i + 1]:
                resultado[i], resultado[i + 1] = resultado[i + 1], resultado[i]
                houve_troca = True

        if not houve_troca:
            break

    return resultado

numeros = [9, 4, 7, 2, 8, 1]

print("Original:", numeros)
print("Ordenada:", bubble_sort(numeros))
```

### 32. Jogo da Velha

Enunciado: Crie um jogo da velha para dois jogadores.

Explicação: Alterna entre X e O e verifica as combinações vencedoras.

Código:


```python
def mostrar_tabuleiro(tabuleiro):
    print()
    print(f"{tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]}")
    print("--+---+--")
    print(f"{tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]}")
    print("--+---+--")
    print(f"{tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]}")
    print()

def venceu(tabuleiro, jogador):
    combinacoes = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    return any(
        tabuleiro[a] == tabuleiro[b] == tabuleiro[c] == jogador
        for a, b, c in combinacoes
    )

tabuleiro = [str(i) for i in range(1, 10)]
jogador = "X"

for rodada in range(9):
    mostrar_tabuleiro(tabuleiro)

    while True:
        posicao = int(input(f"Jogador {jogador}, escolha uma posição (1-9): ")) - 1

        if 0 <= posicao <= 8 and tabuleiro[posicao] not in ("X", "O"):
            break

        print("Posição inválida ou ocupada.")

    tabuleiro[posicao] = jogador

    if venceu(tabuleiro, jogador):
        mostrar_tabuleiro(tabuleiro)
        print(f"Jogador {jogador} venceu!")
        break

    jogador = "O" if jogador == "X" else "X"
else:
    mostrar_tabuleiro(tabuleiro)
    print("Empate!")
```

### 37. Funções Recursivas

Enunciado: Implemente uma função recursiva para calcular o fatorial.

Explicação: A função chama ela mesma até chegar ao caso base.

Código:


```python
def fatorial(n):
    if n < 0:
        raise ValueError("O número deve ser não negativo.")

    if n == 0 or n == 1:
        return 1

    return n * fatorial(n - 1)

numero = int(input("Digite um número: "))

try:
    print(f"{numero}! = {fatorial(numero)}")
except ValueError as erro:
    print(erro)
```

Enunciado: Classifique os números de uma lista como pares ou ímpares.

Explicação: Usa o resto da divisão por 2 para separar pares e ímpares.

Código:


```python
numeros = [10, 7, 3, 8, 12, 15, 20]

pares = []
impares = []

for numero in numeros:
    if numero % 2 == 0:
        pares.append(numero)
    else:
        impares.append(numero)

print("Pares:", pares)
print("Ímpares:", impares)
```

### 46. Calcular Média e Desvio Padrão

Enunciado: Calcule a média e o desvio padrão de uma lista.

Explicação: Calcula a média e depois o desvio dos valores em relação a ela.

Código:


```python
numeros = [10, 12, 9, 11, 13, 8, 14]

media = sum(numeros) / len(numeros)

variancia = sum((numero - media) ** 2 for numero in numeros) / len(numeros)
desvio_padrao = variancia ** 0.5

print(f"Média: {media:.2f}")
print(f"Desvio padrão: {desvio_padrao:.2f}")
```

### 48. Ordenação por Inserção

Enunciado: Implemente o algoritmo Insertion Sort.

Explicação: Insere cada número na posição correta da parte já ordenada.

Código:


```python
def insertion_sort(lista):
    resultado = lista.copy()

    for i in range(1, len(resultado)):
        atual = resultado[i]
        j = i - 1

        while j >= 0 and resultado[j] > atual:
            resultado[j + 1] = resultado[j]
            j -= 1

        resultado[j + 1] = atual

    return resultado

numeros = [9, 5, 1, 4, 3, 8, 2]

print("Original:", numeros)
print("Ordenada:", insertion_sort(numeros))
```

### 50. Cálculo de Frequência de Letras

Enunciado: Calcule a frequência de cada letra em uma string.

Explicação: Percorre o texto e conta quantas vezes cada letra aparece.

Código:


```python
texto = input("Digite um texto: ").lower()
frequencia = {}

for caractere in texto:
    if caractere.isalpha():
        frequencia[caractere] = frequencia.get(caractere, 0) + 1

for letra in sorted(frequencia):
    print(f"{letra}: {frequencia[letra]}")
```
